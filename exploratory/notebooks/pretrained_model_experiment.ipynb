```python
# model_setup.py
import torch, torch.nn as nn, torchxrayvision as xrv

def get_device():
    # Prioridad: MPS (Apple GPU) > CUDA > CPU
    if torch.backends.mps.is_available(): return torch.device("mps")
    if torch.cuda.is_available(): return torch.device("cuda")
    return torch.device("cpu")

def load_pretrained_densenet121_binary():
    """
    DenseNet-121 preentrenado en CXR (TorchXRayVision), cabeza binaria 1 logit.
    """
    model = xrv.models.DenseNet(weights="densenet121-res224-all")
    in_feats = model.classifier.in_features
    model.classifier = nn.Linear(in_feats, 1)
    return model

def freeze_backbone(model, freeze=True):
    for p in model.features.parameters():
        p.requires_grad = not (freeze)
    return model
```
